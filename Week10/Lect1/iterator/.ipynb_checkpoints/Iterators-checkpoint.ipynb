### Intro
**Q))** As you may remember, the two types of for loops in Java,

```java
public class ForEachTester {

	public static void main(String[] args) {
		ArrayList<Integer> list = new ArrayList<Integer>();
		list.add(10);
		list.add(5);
		list.add(20);
		
		// O(N)
		for(int i = 0; i < list.size(); i++) {
			System.out.println(list.get(i));
		}
```

- The first one iterates over the indicies...
  
**Q))** What is the complexity?
- how does get work in an AL? Why is it O(1)?
- The elements are stored contiguously — one after another in memory, just like a plain array.
- There’s no looping, no traversal, no searching.
    - It simply jumps directly to position i in the array.
- arrays in memory support direct indexing.
    - The computer can compute the memory address of any element instantly using this formula:
    - Address of element  𝑖 = Base address + 𝑖 × size of each element
    - This computation takes constant time, no matter how large the array is.

### For-Each ArrayList
- The second iterates over the items in the data collection
- This is what we call for-each loop
    - Here: this is like saying for each int e in list... print e

```java
		// O(N)
		for(int e: list) {
			System.out.println(e);
		}
```

**Q))** What is the complexity?
- This is O(N) because it iterates iecer the list once, everytime retrieving the next item

It looks simple, but Java is actually doing a lot behind the scenes.
- Here's the big reveal:
    - the for-each loop is shorthand that the Java compiler rewrites for you. 
    - Java translates the for-each loop into a regular loop that looks like this

```java
//it uses what we call an iterator behind the scenes.
Iterator<Integer> it = myList.iterator();
//the iterator is like a cursor... 
while (it.hasNext()) { //we check if it has a next item
    int n = it.next(); //we move the iterator the next item
    System.out.println(n); // and print
}

    }//end main
} //end class
```

The for-each loop is just a prettier way to write this pattern! 

### What is an Iterator?
- https://docs.oracle.com/javase/8/docs/api/java/util/Iterator.html
- An Iterator is an object returned from a collection (such as an ArrayList)
    - that you can use to go through the list sequentially.

**[[Check Notes - API Link]]**
- 2 abstract methods that need to be implemented for this interface: hasNext(), next()


```java
// Returns true if the iteration has more elements. 
// (In other words, returns true if next() would return an element rather than throwing an exception.)
public boolean hasNext();

// Returns the next element in the iteration.
// Throws the NoSuchElementException if the iteration has no more elements
public Object next();
```

- So, the for-each loop:
    - Asks the data collection for an Iterator (by calling its iterator() method).
    - Uses that Iterator to repeatedly call:
        - hasNext() — checks if there’s another item
        - next() — gets the next item
That’s it!
👉 The loop keeps going until hasNext() returns false.

- This may look familiar:
    -  It is very similar to how we read in lines from a file - check the notes on files.
- So: The for-each loop: is a Java shortcut for using something called an Iterator.

# Iterable Interface

Now let's understand the players involved.
- There is another interface involved in this (Iterable)
- The built-in data structures that we use for-each with
    - implement something called the iterable interface.
    - Let's check ArrayList and LinkedList API

**[[Look AT API]]**
- Now let's check the iterable Interface (found at the iterator)
- which requires only one abstract method iterator().
- In Java, the relationship between the Iterable interface and the Iterator interface is what allows objects to be used in enhanced for-loops (for-each loops).
    - How are iterable and iterator related and why are they needed for the for-each loop.

**Analogy [[ Slides ]]**
- The Book (Iterable):
    - Contains all the content (your data)
    - Can create bookmarks for different readers
        - Multiple people can read the same book simultaneously with their own bookmarks
- The Bookmark (Iterator):
    - Keeps track of your current position in the book
        - Knows what page you're on
    - Can tell you if there are more pages (hasNext())
    - Moves forward as you read (next())
    - Each reader has their own bookmark, so positions don't interfere
- So
    - Iterable → “I can give you a bookmark (iterator) for this book.”
    - Iterator → is the bookmark:
        - “I can tell you if there’s another page, and I can show it to you.”
        - "I am a bookmark that keeps track of where we are in the collection."

```java
list.iterator(); //returns an object that implements the iterator interface.
```

# The Roster Example

🧱 Making your own class work with for-each
- The class for the data structure you want to iterate over needs to have two things:
- Step 1️⃣ — Implement Iterable
    - We’ll say our class is iterable by adding implements Iterable<T>.
- Step 2️⃣ — Return an iterator
    - We must return an object that knows how to iterate through our data.

**Step 1**
- requires implementing one method: iterator 
- This method simply returns an Iterator for the data structure.
- An Iterator is an object that knows how to go through the elements one at a time.

```java
// Returns an iterator over elements of type T.
public Iterator<T> iterator();
```

### Roster CLass
Consider the following Roster class that uses a basic array to store people. It should be fairly straight-forward:

```java
public class Roster { 
	private final static int MAX_STUDENTS = 40;
	private Student[] theStudents; //uses an array
	private int cnt;
	
	public Roster() {
		theStudents = new Student[MAX_STUDENTS];
		cnt = 0;
	}

    /**
     * Get the number of people in the roster
     * 
     * @return The size of the roster
     */
	public int size() {
		return cnt;
	}

    /**
     * Add a person to the roster
     * 
     * @param p The person to add to the roster
     */
	public void add(Student s) {
		if (cnt < MAX_STUDENTS) {
			theStudents[cnt++] = s;
		}
	}

     /**
     * Get a person from the roster based on index
     * 
     * @param i The index of the person to get
     * @return The person at that index or null if the index is not valid
     */
	public Student get(int idx) {
		return theStudents[idx];
	}

```

### Adding Iterator

If I want to be able to use an Iterator to properly go over the Roster, 

Two Steps: 
1) Tell Java that objects of this class can be looped over usig for-each
2) Tell Java how to iterate over them

For the first step, we will use the Iterable interface

```java
public class Roster implements Iterable<Student> { //Iterable = says “I can be looped over.”
```

then I need to add a method that returns an Iterator. I’ll add the following new method to my Roster class:

```java
public Iterator<Person> iterator() {
    // What do I do now?
}
```

Remember, this method needs to return an Iterator that operates over my Roster.

This brings us to the second step: Telling Java how to iterate over objects of this class.

- Given that Iterator is an interface, that means I need to declare a new class that implements that interface and then return an instance of one.
- So, let’s create a new, inner class that implements the Iterator interface:
    - hasNext() checks before each next() call.
        - checks if we are currently pointing at a valid position
        - if yes we call next()
    - next() gives us the item at the current position,
        - then increments the pointer/tracker.

```java

	private class RosterIterator implements Iterator<Student> { //Iterator = says “Here’s how to loop over me.”
		
		// Pointer to the current element the iterator is pointing to
        private int cur = 0;

        // Return true if there is another element, and false otherwise.
        public boolean hasNext() {
            return (cur < end);
        }

        // Return the next item in the Roster and then increment cur
        // If there is no next item, then throw an exception.
        public Person next() {
            if (hasNext()) {
                return studentList[cur++];
            } else {
                throw new NoSuchElementException();
            }
        }

        // I decided not to implement this, so we throw an appropriate exception.
        public void remove() {
            throw new UnsupportedOperationException();
        }	
	}
```

This class is declared inside of the Roster class. When we declare one class inside of another, we call it an inner class. 
- One advantage of inner classes is that they can access all of the instance variables and methods of the class they are inside of.
- Here, for example, RosterIterator can access both the end and studentList instance variables from the Roster class.
- We also fill in our iterator method from above to return an instance of our new RosterIterator.

```java
    /**
     * Get an iterator for the Roster
     * 
     * @return The iterator
     */

    @Override
	public Iterator<Student> iterator() {
		return new RosterIterator(); //here is the object that will be used to iterate over my objects
	}

```

**TEST**
Using this Roster, we can finally do the for-each loop:

```java

	public static void main(String[] args) {
		Roster r = new Roster();
		r.add(new Student("Ryan", "rdriley"));
		r.add(new Student("Ahmed", "ahmed"));
		r.add(new Student("Fatima", "fsafar"));
		
		//for(int i = 0; i < r.size(); i++) {
		//	System.out.println(r.get(i));
		//}

        //calls iterator() on roster: to create a Bookmark (Iterator)
		for(Student s: roster) {                        // uses the bookmark(iterator) to read through the roster
			System.out.println(s);
		}
	}

} //end ROster class
```

Behind the scenes, this happens:
- Java sees for (Student s : r).
    - It calls r.iterator().
    - It runs hasNext() and next() until done.

**Why Not Only Iterate over the array inside it**
- Your Roster class internally uses an array to store students:
- So inside Roster, you could loop directly over that array: for (Student s : theStudents) { ... }
    - ✅ That works inside the class — because you have direct access to the array.
    - If you try to iterate over roster directly... compile error
        - This does not compile, because Roster itself is not iterable.
    - The for-each loop in Java only works with:
        - Arrays (like Student[]), or
        - Objects that implement the Iterable interface.
- Analogy: Think of it like this:
    - The array is like a bookshelf — you can take books off it directly.
    - The Roster is a classroom — it has a bookshelf inside, but outsiders can’t touch it directly.
    - If you make Roster Iterable, you’re saying:
        - “Here’s a way to go through my bookshelf without opening me up.”

# Iterable/Iterator - MyLinkedList

Java complains if we try to do this... Because our created myLinkedList class is not iterable
		for(String s: sample) {
			System.out.println(s);
		}

**[[Modify the LinkedList implementation:]]**

```java
public class MyLinkedList<ListType> implements Iterable<ListType> {
//...
	@Override
	public Iterator<ListType> iterator() {
		return new LinkedListIterator();
	}
	
	public class LinkedListIterator implements Iterator<ListType> {
		private Node<ListType> tmp = head;

		@Override
		public boolean hasNext() {
			return tmp != null;
		}

		@Override
		public ListType next() {
			if (hasNext()) {
				ListType ret = tmp.getValue();
				tmp = tmp.next;
				return ret;
			} else {
				throw new NoSuchElementException();
			}
		}
		
	}

//...
}
```

```java
public class ForEachLLTester {

    public static void main(String[] args) {
		MyLinkedList<String> sample = new MyLinkedList<String>();
		sample.add("Hi");
		sample.add("There");
		sample.add("Me");
		System.out.println(sample);

        //
		for(String s: sample) {
			System.out.println(s);
		}
	}
```

# Why is This Useful?
So, why is this useful? Here are a few reasons:

- There is a uniform way to go through (iterate over) all of the elements of a variety of data structures.
- For-each is frequently more readable and faster to write.
- For some data structures, having this sort of control over iteration can make things more efficient. Consider a linked list: A traditional for loop going through a linked list and using get(i) is 
O(N2)  because get(i) is O(N)  and is being called from a loop over N items.
    - With a properly written iterator, however, a for-each loop through a linked list can be O(N)

# Key Takeaways

- For-each is syntax sugar: 
    - It's automatically converted to use Iterator behind the scenes
- Iterable = "I can be traversed":
        - Implement this interface and provide an iterator() method
- Iterator = "I track position": 
    - This object knows where we are in the collection and how to get the next element
- The relationship: Iterable creates Iterators. Iterators do the actual traversing.

- Why this design?: It separates the collection (Playlist) from the traversal logic (PlaylistIterator). 
    - This is a design pattern called the Iterator Pattern.
