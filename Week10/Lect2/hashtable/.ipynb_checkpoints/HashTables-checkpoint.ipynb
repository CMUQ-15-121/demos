# Intro

- Hash tables are unordered data structures. It stores items, unordered, into an array in such a way that adding new items and finding existing items is extremely fast.
- designed for extremely fast add and find operations.
- Their efficiency, in practices, approaches O(1).
    - Although the theoretical worst-case efficiency of the hash table presented here will be 
O(N), in practice hash tables are much closer to O(1) 
- Wait... if the items are stored unordered, how can we find them quickly?
    - Don't we need to search all of them? (linear search and is therefore O(N))?
- We're going to use a weird trick.

### Hash Function

Let's define a new concept called a hash function. 
- A hash function takes an item (like a number, a string, a Person object, etc) and reduces it to a single number.
- A hash function has the following properties:
    - It must always return the same number for the same item. (It is deterministic.)
    - It should produce a uniform distribution of values. 
        - This means that, roughly, speaking, if you hash 1,000,000 different items, you should not get that many items that have the same hash.

#### Hash Functions in Java: 
In Java, objects provided by the Java API provide their own hash functions in the hashCode() method.
- You can call .hashCode() on most objects and it will return an integer that is the hash of that object.
-  If you write your own object, then you should provide your own hashCode() method. (More on this later.)  

### Example: Hash Function for Strings

Imagine that I want a hash function for lowercase words:

- I will map letters to numbers. a = 0, b = 1, c = 2, etc.
- I will sum the letters for the word. The resulting number is my hash.

Here is some basic java code (have this ready to show, don't write it):
🧠 What it does
- Starts with res = 0.
- For every character in the string:
    - Gets its position in the alphabet ('a' → 0, 'b' → 1, 'c' → 2, etc.).

Adds that value to res.

````java
public class HashFunctionDemo {
	public static int hashString(String s) {
		int res = 0;
		for(int i = 0; i < s.length(); i++) {
			char c = s.charAt(i);
			int val = (int)c - (int)'a';
			res += val;
		}
		return res;
	}
	
	public static void main(String[] args) {
		System.out.println(hashString("qatar"));
	}
}
````

Let's do an example: qatar.
```
q = 16
a = 0
t = 19
a = 0
r = 17
Sum is 52. So that's my hash for Qatar.
```

**Analysis of the example**
Q)) Does this meet my properties?
- It does always return the same number for the same item.
- It is not very uniformly distributed.
    - Similar words will have similiar hashes, lots of words will have the same hash, etc.
    - ⚠️ Why this is a bad hash function
        - This approach just adds values, ignoring the order of letters.
        - "artq" would also give 52!
        - (same letters, just rearranged)
        - That means lots of collisions — different words get the same hash value.
      
In reality, getting a uniform distribution is complicated.
- So I'm going to go to a math person and ask them about it.
- And they tell me that using prime numbers and doing something like this is better:
    -  (have this ready to show, don't write it)

````java
public class BetterHashFunctionDemo {
	public static int hashString(String s) {
		int prime = 31;
		int res = 1;
		for(int i = 0; i < s.length(); i++) {
			char c = s.charAt(i);
			int val = (int)c - (int)'a'; //Ascii code for chars
			res = res * prime + val; 
		}
		return res;
	}


    //first iter, the val for letter q gets added to 31...
    //second iter, the prev result (which included q) gets mul;tipled by 31 + the val for etetr a
    //thirs iter, the prev resilt (which includes q and a) gets multiple by 31 again + the val for t
        //and so on.
    //so the number of times each letter ends up being multipled by 31, depends on its position in the word
        //earlier letters get multiople more by 31
    //This is what makes this approach sentivie to letter position in the word.
        //if we change the letters order, it will produce a different hash value
	
	public static void main(String[] args) {
		System.out.println(hashString("qatar"));
	}
}
````

uses a polynomial hash with a prime multiplier (31), 
- producing unique, well-spread values that depend on both the characters and their positions.
- Instead of just adding, it builds a weighted sum using powers of a prime number (31):
    - This means each letter’s position matters — earlier letters contribute much more weight.
    - Because each new loop multiplies the entire running total by 31 again.
        - Every new character multiplies the previous total by 31,
            - so earlier letters keep getting multiplied more times →
            - earlier letters have higher powers of 31 in the final sum. 
        - That pushes the contributions from earlier characters one level higher each time.
        - So if your string has n characters:
            - The first letter is multiplied by 31^𝑛−1 times
            - The second by 31 ^ 𝑛−2 times
            - The last by 31 ^ 0...
        - Before each char regardless of its position in the string had the same value
- Multiplying by a prime number mixes up the values better,
    - making small changes in the string (like swapping letters)
    - produce very different hash results —
    -  which is exactly what you want in a good hash function.

# Hash Code

In Java, most objects implement their own hash function suitable for calculating their hash. 
- This is the hashCode method.
- Java provided classes have a built-in method that can calculate the hash of that object: hashCode.

````java
public class HashcodeTester {

	public static void main(String[] args) {
		String s = "Hi there";
		String s2 = "Hi therf";

		System.out.println(s.hashCode());
		System.out.println(s2.hashCode());
	}

}
````

### Creating Your Own

If you are writing a class that needs to be able to be hashed, you need to provide a good hashCode method.
- If you write your own object,
    - it will include a default hashCode method that will work just fine.
- However, if your object provides a custom compareTo() method and/or an equals() method,
    - then you must provide your own hashCode method that is consistent with them.
    - In this case, consistent means that if compareTo() and equals() say that two objects are equal,
    - then their hashes must be the same as well.
    - In practice, this usually just means generating an object’s hash codes based on the hash codes of its attributes.

### Inconsistency Issue

````java
public final class Student {
	  private final String name;
	  private final int id;

	  public Student(String name, int id) { this.name = name; this.id = id; }

	  @Override public boolean equals(Object o) {
	    if (this == o) return true;
	    if (!(o instanceof Student s)) return false;
	    return this.id == s.id;              // equality by ID
	  }

	  @Override public int hashCode() {
	    return name.hashCode();              // ❌ hashing by NAME (inconsistent!)
	  }
	}
}
````

What goes wrong in a hash table

Two students with same ID but different names are equal per equals(), but produce different hash codes, so they land in different buckets and both remain in the table.

Students:
  s1 = Student("Alice", 100)   // equals compares ID → s1 == s2
  s2 = Student("Bob",   100)

Hash codes (wrongly from name):
  hash("Alice") % 4 -> 1
  hash("Bob")   % 4 -> 3

Buckets:
 index 0:  [  ]
 index 1:  [ s1("Alice",100) ]
 index 2:  [  ]
 index 3:  [ s2("Bob",100)   ]


Good (consistent) version

If id defines identity, use it for both equals() and hashCode():

````java
@Override public boolean equals(Object o) {
  if (this == o) return true;
  if (!(o instanceof Student s)) return false;
  return this.id == s.id;              // equality by ID
}

@Override public int hashCode() {
  return Integer.hashCode(id);         // ✅ hashing by the same field (ID)
}
````

Now both students with ID 100 hash the same, go to the same bucket, and the second insert is recognized as a duplicate (or replaces the value, if you’re a HashMap).

### Guidelines for a HashCOde() and equals()
    - SUmmary
        - Have to use the same attributes for hashCode(_) and equals()
        - Using primes helps achieve better even distribution
